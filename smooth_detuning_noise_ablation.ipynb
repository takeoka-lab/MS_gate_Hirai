# Smooth-detuning MS gate: motional-noise ablation

This notebook separates the standard-to-smooth change into the noise-free QPT floor, heating, motional dephasing, their non-additive interaction, and the remaining spin/optical/cross contribution. Reusable functions are defined in `smooth_detuning_analysis.py`.

```python
from pathlib import Path
import importlib

from IPython.display import Image, display

import smooth_detuning_analysis as sda
sda = importlib.reload(sda)

NBAR_VALUES = (0.01, 1.0, 2.0, 3.0, 4.0)
NOISE_CASES = (
    'none',
    'heating_only',
    'motional_dephasing_only',
    'all_motional',
    'all_noise',
)
EXECUTE_QPT = True
RESUME = True
PARALLEL_WORKERS = 1
OUTPUT_DIR = Path('results/smooth_detuning_noise_ablation')
ALL_NOISE_SOURCE = Path('results/smooth_detuning_nbar_sweep')

SWEEP_CONFIG = sda.NbarSweepConfig(
    nbar_values=NBAR_VALUES,
    noise_case='all_noise',
    scattering_model='constant_rate',
    standard_time_points=501,
    smooth_time_points=1501,
    standard_solver_max_step_s=0.2e-6,
    smooth_solver_max_step_s=0.05e-6,
    parallel_workers=PARALLEL_WORKERS,
    show_progress=True,
)
SWEEP_CONFIG
```

## Run or resume

The completed `all_noise` sweep is reused from `results/smooth_detuning_nbar_sweep`. The other four cases are calculated and saved independently for every pulse/$\bar{n}$ condition.

```python
RESULT = sda.run_noise_ablation(
    output_dir=OUTPUT_DIR,
    sweep_config=SWEEP_CONFIG,
    noise_cases=NOISE_CASES,
    all_noise_source_dir=ALL_NOISE_SOURCE,
    execute_qpt=EXECUTE_QPT,
    resume=RESUME,
)
print('Output directory:', RESULT['output_dir'].resolve())
```

## Component results and attribution

Positive `smooth_advantage = I_standard - I_smooth` means that the component favors smooth detuning. The decomposition retains a separate motional interaction term because Lindblad infidelities are not generally additive. The spin/optical remainder also contains cross terms and must not be interpreted as a pure scattering rate.

```python
ablation = RESULT['ablation_summary']
attribution = RESULT['attribution']
residual = RESULT['residual_spin_motion']

print('First-order final residual-displacement component:')
display(residual)
display(Image(filename=str(OUTPUT_DIR / 'figures' / 'residual_spin_motion_infidelity.png')))

if attribution.empty:
    print('Ablation QPT is incomplete. Leave RESUME=True and rerun this notebook.')
    if not ablation.empty:
        display(ablation[['pulse_type', 'noise_case', 'nbar', 'infidelity']])
else:
    display(attribution.pivot(index='nbar', columns='mechanism', values='smooth_advantage'))
    display(Image(filename=str(OUTPUT_DIR / 'figures' / 'noise_ablation.png')))
    display(Image(filename=str(OUTPUT_DIR / 'figures' / 'motional_noise_penalties.png')))
    display(Image(filename=str(OUTPUT_DIR / 'figures' / 'mechanism_attribution.png')))
```

## Output files

- `ablation_summary.csv`: exact QPT Infidelity for every pulse/noise/$\bar{n}$ condition.
- `mechanism_attribution.csv`: signed components that sum to the all-noise smooth advantage.
- `residual_spin_motion.csv`: first-order $\alpha(T)$-only component and angle-only component.
- `conditions/<noise_case>/`: resumable per-condition QPT outputs and channel NPZ files.
- `figures/`: ablation, noise-penalty, residual-displacement, and attribution plots.

## Individual spin-dephasing and photon-scattering errors

The existing motional-noise ablation above is left unchanged. This additional measurement runs two single-noise QPT sweeps using the same pulse, $\bar{n}$ grid, solver settings, and `constant_rate` scattering model:

- `spin_dephasing_only`: $T_2^*=0.3$ s; all other noise rates are zero.
- `photon_scattering_only`: the existing Rayleigh ($3\,\mathrm{s}^{-1}$) and Raman ($1\,\mathrm{s}^{-1}$) channels are enabled together; heating, motional dephasing, and spin dephasing are zero.

For each pulse and $\bar{n}$, the reported individual error is $I_{\mathrm{only}}-I_{\mathrm{none}}$, so the matching noise-free QPT floor is removed. These single-noise increments exclude cross terms and therefore need not sum to the `all_noise` Infidelity.

```python
SPIN_OPTICAL_RESULT = sda.run_spin_optical_individual_ablation(
    output_dir=OUTPUT_DIR,
    baseline_summary=ablation,
    sweep_config=SWEEP_CONFIG,
    noise_cases=sda.SPIN_OPTICAL_INDIVIDUAL_CASES,
    execute_qpt=EXECUTE_QPT,
    resume=RESUME,
)
print('Spin/optical output directory:', SPIN_OPTICAL_RESULT['output_dir'].resolve())
```

```python
sda = importlib.reload(sda)
spin_optical_errors = SPIN_OPTICAL_RESULT['individual_errors']

if spin_optical_errors.empty:
    print('Individual QPT is incomplete. Leave EXECUTE_QPT=True and RESUME=True, then rerun this cell.')
else:
    columns = [
        'noise_case',
        'nbar',
        'standard_noise_only',
        'standard_none',
        'standard_individual_error',
        'smooth_detuning_noise_only',
        'smooth_detuning_none',
        'smooth_detuning_individual_error',
        'smooth_error_reduction',
        'standard_over_smooth_error',
    ]
    display(spin_optical_errors[columns])
    individual_figure = sda.plot_spin_optical_individual_errors(
        spin_optical_errors,
        OUTPUT_DIR / 'figures' / 'spin_optical_individual_errors.png',
    )
    display(Image(filename=str(individual_figure)))
    print('Positive smooth_error_reduction means the individual noise error is smaller for smooth detuning.')

print('Saved files:')
print('-', OUTPUT_DIR / 'spin_optical_individual_summary.csv')
print('-', OUTPUT_DIR / 'spin_optical_individual_errors.csv')
print('-', OUTPUT_DIR / 'spin_optical_individual_config.json')
```
